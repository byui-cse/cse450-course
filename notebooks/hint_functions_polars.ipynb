```
import polars as pl

# Create a data frame to test with
data = pl.DataFrame({"celsius": [10,20,30,40], "weather": ['sun', 'sun', 'rain', 'rain']})
data
```

```
# Define a function that takes one column and returns the conversion
def convert_to_f(col):
  return (col * 9/5) + 32

# Apply the function to each element in the celsius column
data = data.with_columns(convert_to_f(pl.col('celsius')).alias('fahrenheit'))
data
```

```
# Do this with native expressions instead.
#
# There is no real advantage to using one way over the other
data = data.with_columns(((pl.col('celsius') * 9 / 5) + 32).alias('fahrenheit'))
data
```

```
# Define a function that expects a row of data and calculates a result based on
# different columns in that row:
def should_go_to_beach(weather, celsius):
  # We will only go to the beach if it's sunny and warm
  return (weather == 'sun') & (celsius >= 20)

# Here we apply the function to the entire DataFrame rather than a single
# column
data = data.with_columns(should_go_to_beach(pl.col('weather'), pl.col('celsius')).alias('beach_trip'))
data
```

```
# And an inline version:
data = data.with_columns(((pl.col('weather') == 'sun') & (pl.col('celsius') >= 20)).alias('beach_trip'))
data
```
